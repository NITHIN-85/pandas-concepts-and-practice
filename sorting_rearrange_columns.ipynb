```python
#creating a MultiIndex DataFrame
index = pd.MultiIndex.from_tuples([('a',1),('b',2),('a',2),('b',1)])
df_multi=pd.DataFrame({'A':[1,2,3,4]},index=index)
print("original Multiplex DataFrame:")
df_multi
```

```python
#sorting by the first level of the index
sorted_Multi = df_multi.sort_index(level=0)
print("\nMultiIndex DataFrame after sorting by the first level of index:")
sorted_Multi
```

```python
#sorting by the first level of the index
sorted_Multi_level2 = df_multi.sort_index(level=1)
print("\nMultiIndex DataFrame after sorting by the first level of index:")
sorted_Multi_level2
```

#Rearrange order of Columns

```python
import pandas as pd
#Example DataFrame
df=pd.DataFrame({
    'B':[1,2,3],
    'A':[4,5,6],
    'C':[7,8,9]
})
print("original DataFrame:")
df
```

```python
#specifying the new order of columns
new_order =['A','B','C']
```

```python
#Rearrange the columns
df_recorded = df[new_order]
print("DataFrame after rearranging columns:")
print(df_recorded)

```

```python
df_recorded = df.reindex(['A','B','C'],axis=1)
df_recorded
```

```python
#Alphabetically
df_sorted = df[sorted(df.columns)]

#Custom sort (e.g.,reverse alphabetical order)
df_custom_sorted = df[sorted(df.columns,reverse=True)]
df_custom_sorted
```

```python
df_reordered = df.loc[:,['A','B','C']]
df_recorded
```

```python
#Assuming you know the position of the columns you want to reorder
df_recorded = df.iloc[:,[1,0,2]] #Here,1,0,2 represents the new order of column indices
df_recorded
```
