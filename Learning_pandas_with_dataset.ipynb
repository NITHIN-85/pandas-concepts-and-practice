```python
#Droping or Deleting the DataFrames
```

```python
df = pd.read_csv('Iris.csv')
df
```

```python
# Dropping columns by using names.
df_dropped_columns=df.drop(columns=['SepalWidthCm'])
df_dropped_columns
```

```python
# Dropping rows by using indexes.
df_dropped_rows = df.drop([0,2])
df_dropped_rows
```

```python
df
```

```python
# modify the original DataFrame
df.drop(columns=['SepalWidthCm'],inplace=True)
df
```

```python
# Drop rows by index and modify the original DataFrame
df.drop([0,2],inplace=True)
df
```
