```python
# ILOC Function : it is used to access the rows and columns by their integer position. It is similar to loc function but it uses integer index instead of actual values of index and columns.
# This is number based, not the actual values
#ILOC means index based data selecting method. 
```

```python
df=pd.read_csv('Iris.csv')
#setting any column of df as index
df.set_index('SepalWidthCm',inplace=True)
df
```

```python
df1
```

```python
df1.iloc[0] # it will access the first row of the DataFrame
```

```python
df1.iloc[5] # it will access the 5th row of the DataFrame
```

```python
df.iloc[[0,1]] # it will access the rows with indices 0 and 1
```

```python
df.iloc[:3]# it will access the first 3 rows of the DataFrame
```

```python
df.iloc[lambda x: x.index %2==0]#it will access the rows with even index values of the DataFrame
```

```python
df.iloc[1:3,0:3] # it will access the rows with index values 1,2,3 of the DataFrame
```

```python
df.columns
```

```python
df.iloc[1:5,1:4] # it will access the rows with index values 1,2,3,4 and columns with index values 1,2,3 of the DataFrame
```

```python
df.iloc[2,2]
```

```python
df.iloc[2,2]=100 # modify the value of the cell at row index 2 and column index 2 to 100
```
